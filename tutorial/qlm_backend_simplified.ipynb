## Imports and Setup

First, let's import OpenVQA and see how to access both VQE and QAOA.

```python
import os
import sys

# Ensure project root on path
sys.path.insert(0, os.path.abspath('../..'))



# Import the main engine and algorithms
from openvqa import VQA_myQLM
from openvqa.algorithms import VQE, QAOA

print(' OpenVQA imports ready')
print('  - VQA_myQLM: Main execution engine')
print('  - VQE: Variational Quantum Eigensolver')
print('  - QAOA: Quantum Approximate Optimization Algorithm')
```

```python
# Print supported options exposed by OpenVQA (ordered output)
from openvqa.options import AVAILABLE_OPTIONS

vqe = AVAILABLE_OPTIONS.get('vqe', {})

print('FEATURE MAPS')
print(vqe.get('feature_maps', []))

print('\nANSATZ')
print(vqe.get('ansatz', {}))

print('\nOPTIMIZERS')
print(vqe.get('optimizers', []))
```

```python
# VQE run WITH shots (sampling noise)
import importlib
import openvqa.algorithms.vqe as vqe_mod
importlib.reload(vqe_mod)
from openvqa import VQA_myQLM
from openvqa.algorithms.vqe import VQE

engine = VQA_myQLM(
    n_qubits=2,
    backend='qat.simulators.qubit',
    shots=1000,
    seed=0,
    verbose=True,
 )

vqe = VQE(
    hamiltonian='H2',
    ansatz='StronglyEntanglingLayers',  # PennyLane template name
    embedding=None,
    feature_map='ZZFeatureMap',          # Qiskit feature map name
    optimizer='cobyla',
    max_iterations=30,
    n_layers=1,
 )

results = engine.run(vqe)
print('\nVQE energy:', results.energy)
```

```python
# QAOA run WITH shots (sampling noise)
import importlib
import openvqa.algorithms.qaoa as qaoa_mod
importlib.reload(qaoa_mod)
from openvqa import VQA_myQLM
from openvqa.algorithms.qaoa import QAOA

engine = VQA_myQLM(
    n_qubits=4,
    backend='qat.simulators.qubit',
    shots=1000,
    seed=0,
    verbose=True,
 )

qaoa = QAOA(
    problem='maxcut',
    graph_size=4,
    p=1,
    feature_map='ZZFeatureMap',  # Qiskit feature map name (optional init)
    optimizer='cobyla',
    max_iterations=30,
 )

results = engine.run(qaoa)
print('\nQAOA objective (energy):', results.energy)
```
