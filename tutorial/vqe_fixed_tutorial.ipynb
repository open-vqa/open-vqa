## Available combinations (edit the constants, no prompts)

OpenVQA exposes multiple choices for backends and circuit templates. This notebook uses only one combination, but you can change it by editing the constants in **Cell 3**.

### Backends (engine `backend=`)
- **QLM simulator**: `qat.simulators.qubit` (recommended default)
- **Qiskit Aer**: `aer_simulator` (requires Qiskit Aer installed)
- **PennyLane path**: any string containing `pennylane` triggers the algorithm's PennyLane execution path (it uses `default.qubit` internally).
- **PennyLane templates → QLM**: keep backend as QLM and set `use_pennylane_via_qlm=True` when initializing the engine.

### VQE ansatz choices (algorithm `ansatz=`)
- Built-in: `ryrz`, `linear`, `uccsd`
- Qiskit library: `RealAmplitudes`, `EfficientSU2`, `TwoLocal`
- PennyLane templates (via conversion or PennyLane execution): `StronglyEntanglingLayers`, `BasicEntanglerLayers`, `RandomLayers`

### Feature maps (algorithm `feature_map=`; Qiskit circuits)
- `PauliFeatureMap`, `ZZFeatureMap`, `ZFeatureMap`

### Embeddings (algorithm `embedding=`; PennyLane-style data encoding)
- `AngleEmbedding`, `AmplitudeEmbedding`, `BasisEmbedding`, `DisplacementEmbedding`

**Compatibility note**: `feature_map` is intended for Qiskit circuits; `embedding` is intended for PennyLane-style templates. Typically you use one or the other (not both).

```python
import os
import sys
import numpy as np

# Ensure project root is on sys.path (so import openvqa works when running from openvqa/tutorial)
sys.path.insert(0, os.path.abspath('../..'))

# IMPORTANT: force Python to use this workspace's OpenVQA code
for name in list(sys.modules.keys()):
    if name == 'openvqa' or name.startswith('openvqa.'):
        del sys.modules[name]

from openvqa import VQA_myQLM
from openvqa.algorithms import VQE
import openvqa as _openvqa

print('✓ OpenVQA imported from:', os.path.dirname(_openvqa.__file__))
print('  NumPy:', np.__version__)
```

```python
# ------------------------
# Fixed tutorial settings
# ------------------------
N_QUBITS = 2
BACKEND = 'qat.simulators.qubit'
HAMILTONIAN = 'H2'
FEATURE_MAP = 'ZZFeatureMap'
ANSATZ = 'RealAmplitudes'
OPTIMIZER = 'cobyla'
MAX_ITERATIONS = 50
N_LAYERS = 2

print('Configuration:')
print('  backend      =', BACKEND)
print('  hamiltonian  =', HAMILTONIAN)
print('  feature_map  =', FEATURE_MAP)
print('  ansatz       =', ANSATZ)
print('  optimizer    =', OPTIMIZER)
print('  max_iters    =', MAX_ITERATIONS)
print('  n_layers     =', N_LAYERS)
```

```python
engine = VQA_myQLM(
    n_qubits=N_QUBITS,
    backend=BACKEND,
    shots=1000,
    seed=42,
    verbose=False,
)
print('✓ VQA Engine initialized')
```

```python
vqe = VQE(
    hamiltonian=HAMILTONIAN,
    ansatz=ANSATZ,
    feature_map=FEATURE_MAP,
    optimizer=OPTIMIZER,
    max_iterations=MAX_ITERATIONS,
    n_layers=N_LAYERS,
)
print('Running VQE...')
results = engine.run(vqe)
print('Energy:', results.energy)
print('Time (s):', results.wall_time)
```
