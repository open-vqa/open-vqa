# Set up the OpenVQA environment

Do this once per computer. **Part 1** creates a Python environment from a terminal.
**Part 2** runs inside this notebook and checks that everything works.

**You need:** Python 3.10, 3.11 or 3.12 (myQLM has no builds for 3.13 or newer),
Git, and an internet connection. The install downloads about 170 packages, so expect
5 to 15 minutes.

*Tested on Windows with Python 3.10 and 3.12.*

## Part 1: create the environment (in a terminal)

### Option A: conda (Anaconda or Miniconda)

On Windows, open **Anaconda Prompt**.

```bash
conda create -n openvqa python=3.12 -y
conda activate openvqa
git clone https://github.com/open-vqa/open-vqa.git
cd open-vqa
git checkout alpha1.0
python -m pip install -r requirements.txt
```

### Option B: plain Python with venv

Windows (PowerShell):

```bash
git clone https://github.com/open-vqa/open-vqa.git
cd open-vqa
git checkout alpha1.0
py -3.12 -m venv .venv
.venv\Scripts\activate
python -m pip install -r requirements.txt
```

`.venv` is the name of the environment folder. If you choose another name, use the
same name in the `activate` command.

macOS / Linux (not tested yet; myQLM publishes builds for both):

```bash
git clone https://github.com/open-vqa/open-vqa.git
cd open-vqa
git checkout alpha1.0
python3.12 -m venv .venv
source .venv/bin/activate
python -m pip install -r requirements.txt
```

**No Git?** On the GitHub page, switch the branch selector to **alpha1.0**, click
**Code → Download ZIP**, unzip it, and `cd` into the unzipped folder instead of
running `git clone` and `git checkout`.

`git checkout alpha1.0` selects the branch that contains OpenVQA; the `main` branch does
not have it yet.

`requirements.txt` installs the tested version of every library, plus OpenVQA itself.
The last line of the output should start with `Successfully installed`.

- **VS Code:** open the `open-vqa` folder, open this notebook, click **Select Kernel**

## Part 2: check the installation

Run the three cells below from top to bottom.

````python
import sys

print("Python:", sys.version.split()[0])
print("Interpreter:", sys.executable)
assert (3, 10) <= sys.version_info[:2] <= (3, 12), "Use Python 3.10, 3.11 or 3.12"
````

````python
from importlib.metadata import version, PackageNotFoundError

for name in ["openvqa", "qiskit", "myqlm", "myqlm-interop", "pennylane",
             "numpy", "scipy", "networkx", "matplotlib"]:
    try:
        print(f"{name:15s} {version(name)}")
    except PackageNotFoundError:
        print(f"{name:15s} NOT INSTALLED")
````

````python
import openvqa
from openvqa import VQA_myQLM, VQE

print("OpenVQA", openvqa.__version__, "from", openvqa.__file__)
result = VQA_myQLM(n_qubits=2, seed=42).run(
    VQE(hamiltonian="H2", optimizer="cobyla", max_iterations=300))
print("VQE energy:", round(result.energy, 6), "(expected: -0.5)")
````

**Expected:** the first cell shows Python 3.10, 3.11 or 3.12 and the path of your
new environment; the second lists a version for every library (no `NOT INSTALLED`);
the third prints **VQE energy: -0.5**. Then you are ready: continue with
`vqe_tutorial.ipynb`, then `qaoa_tutorial.ipynb`.

## Troubleshooting

| Problem | What to do |
|---------|------------|
| `No module named 'openvqa'` (or `qat`) | The notebook kernel is a different Python. Select the environment you created (see above). |
| `No matching distribution found for qat-core` or `requires a different Python` | Your Python is 3.13 or newer. Create the environment with Python 3.12. |
| `conda` is not recognised | On Windows, use **Anaconda Prompt** instead of a normal terminal. |
| PowerShell refuses to run `activate` | Run `Set-ExecutionPolicy -Scope CurrentUser RemoteSigned` once (it allows local scripts), then try again. |
| `pip install -r requirements.txt` installs nothing | You are on the `main` branch. Run `git checkout alpha1.0`, then install again. |
| The install stops with a network error | Run the same `pip install` command again; finished downloads are reused. |
